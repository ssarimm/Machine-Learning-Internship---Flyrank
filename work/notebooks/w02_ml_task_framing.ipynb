## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row = one page (content_id), belonging to one client.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

cols = ["impressions_90d", "days_since_last_update", "avg_position", "ctr",
        "word_count", "is_declining_label"]
slice_df = df[cols]

print("Rows (pages):", len(slice_df))
print("Unique pages:", df["content_id"].nunique(), "| clients:", df["client_id"].nunique())
print("Declining rate (random-pick baseline for Precision@K):",
      round(df["is_declining_label"].mean(), 3))
slice_df.head()
```
